##### Handling negative values in usage/activity columns

Several columns that measure hours or counts contained negative values, which isn't logically possible.

Two possible causes were considered:
- **Accidental entry error** (e.g. a typo where `1.55` was entered as `-1.55`) — if this were the case, the correct fix would be to take the absolute value, restoring the original intended number.
- **Synthetic data generation noise** — since both datasets are synthetic (not collected from real people), values were likely generated from a statistical distribution with no built-in floor at 0, allowing some values to fall slightly below zero at random.

Given the fact that the negative values also appear in unrelated columns (e.g. `Income_USD` as low as -25,622, `Push_Notifications_Per_Day` as low as -21), this pattern is more consistent with generator noise than data-entry error. A typo explanation is unlikely to produce negative values that large or that widespread.

Based on this reasoning, negative values in usage/activity columns were treated as invalid and clipped to a minimum of 0 using `.clip(lower=0)`, rather than corrected.

```python
# Clipping the negative values to zero

columns_to_clip = ['Income_USD','Daily_Usage_Hours','Phone_Checks_Per_Day',
                    'Time_on_Social_Media', 'Time_on_Gaming', 'Streaming_Usage_Hours', 
                    'Messaging_Usage_Hours','Work_Related_Usage_Hours','Exercise_Hours',
                    'Time_Spent_With_Family_Hours','Online_Shopping_Hours','Monthly_Data_Usage_GB',
                    'Push_Notifications_Per_Day'
                    ]
df[columns_to_clip] = df[columns_to_clip].clip(lower=0)
```

##### Changing the values of Addiction_level to Numeric to match with Dataset 1

- Dataset 2 - (global_smartphone_addiction) : Addiction_Level is categorical - Low/Medium/high/Severe 

Mapping - Addiction_Level

    Low         -       2.5 (mid range 1-4) 
    Moderate    -       5.5 (mid range 5-6)
    High        -       7.5 (mid range 7-8)
    Severe      -       9.5 (mid range 9-10)

##### Changing Values of the column named User_ID

**Both Datasets use their own User_ID columns, numbered from 1 to 3000**
 
 This is will create duplicates during the merging process. To fix this issue, global_smartphone_addiction dataset's User_IDs will offset by adding 10,000 to the current values - shifting its range to 10001–13000

```python
# Using an if statement to check whether User_ID values don't keep increasing every time "Run All" is used.
# Offset Dataset 2's User_IDs to avoid duplicates with Dataset 1 during the merge

if df['User_ID'].max() < 10000:
    df['User_ID'] = df['User_ID'] + 10000

df['User_ID'].head()
```

### Step 3 - Data Visualization

**Scatter plot to view the relationship of Sleep_Hours, Addiction_Level and Anxiety_Level**

The scatter plot shows no visible relationship between `Sleep_Hours` and `Anxiety_Level`. Points are spread evenly across the full range of both axes, with no upward or downward trend, and all four addiction levels (2.5, 5.5, 7.5, 9.5) appear mixed throughout the chart rather than occupying distinct regions. People sleeping as little as 3 hours and as much as 9+ hours both show anxiety levels spanning the entire 0–10 range.

This is consistent with the weak correlation (-0.22) observed between `Sleep_Hours` and `Addiction_Level` in Dataset 1 (teen_screentime_addiction), suggesting the lack of a strong relationship may hold across the broader population as well.



```python
# Scatter plot 

fig, ax = plt.subplots(figsize=(10, 8))

sns.scatterplot(data=df, x='Sleep_Hours', y='Anxiety_Level', hue='Addiction_Level', 
                 palette='coolwarm', alpha=0.8, ax=ax)

fig.suptitle('Sleep Hours vs Anxiety Level', fontweight='bold', color="#772222", fontsize=16)
plt.title('(Survey based on 3000 participants)', fontsize=11, color='black', pad=7)
ax.set_xlabel('Sleep Hours', fontsize=13, color="#772222")
ax.set_ylabel('Anxiety Level (1-10)', fontsize=13, color="#772222")

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()
```

#### Heat Map to view the correlation of the variable

Nearly every correlation in this heatmap falls between -0.03 and 0.05, with the only exception being each variable's perfect self-correlation (1.00) on the diagonal. None of the variables show any meaningful linear relationship with one another. This is an unusual result. 

In real-world survey data, it's rare for every single variable pair to be this close to zero — even weak but genuine relationships (like the ones commonly seen between sleep and mental health, or screen time and mood) typically produce correlations in the 0.1–0.3 range at minimum. Therefore, the heatmap does not provide a meaningful support for or against the research questions being investigated. 

```python
# To see the correlation of numeric values

fig, ax = plt.subplots(figsize=(14, 14))

sns.heatmap(df_correlation.round(2), annot=True, fmt=".2f", cmap='coolwarm', linewidths=0.5, annot_kws={'size': 10})
plt.suptitle('Global Smartphone Addiction Correlation Map', fontsize=20, color= "#772222", fontweight = 'bold')
plt.title('(Survey based on 3000 Participants (Age Range 12-70))', fontsize=11, color='black', pad=8)
plt.xticks(rotation=90, ha='center', fontsize=12)
plt.yticks(rotation=0, fontsize=12)

plt.tight_layout()
plt.show()
```

#### Daily Smartphone Usage distribution using histogram

The histogram displays a univariate analysis of daily phone usage. The data appears balanced with no extreme skew. The most common usage is around 6 hours per day, with the majority of students falling within the 4–8 hour range.

```python
#To check the daily usage distribution

fig, ax = plt.subplots(figsize=(12, 12))  
ax.hist(df['Daily_Usage_Hours'], bins=20, alpha=0.9, color= 'steelblue', edgecolor = 'white')

fig.suptitle('Daily Smartphone Usage Distribution', fontweight = 'bold', color= "#772222" , fontsize=20)
plt.title('(Survey based on 3000 participant (Age Range: 12-70 Years))', fontsize=11, color='black', pad=8)
ax.set_xlabel('Daily phone usage (hours)', fontsize=15,fontweight = 'bold', color= "#772222" )  
ax.set_ylabel('Participant count', fontsize=15,fontweight = 'bold', color= "#772222" )         


mean_value = df['Daily_Usage_Hours'].mean()
ax.axvline(mean_value, color='red', linestyle='--', linewidth=1.5, label=f'Mean: {mean_value:.1f} hrs')
ax.legend()

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show()

```
